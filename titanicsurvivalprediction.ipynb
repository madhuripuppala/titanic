```python
import numpy as np
import pandas as pd
```

```python
import io
data = pd.read_csv(io.BytesIO(uploaded['tested.csv']))
```

```python
data.head()
```

```python
women = data.loc[(data["Sex"] == "female") & (data["Survived"] == 1)]
rate_women = women["Survived"].mean()
print("% of women who survived:", rate_women * 100)
```

```python
men_survived = data[(data["Sex"] == "male") & (data["Survived"] == 1)]
total_men = len(data[data["Sex"] == "male"])
rate_men = (len(men_survived) / total_men * 100) if total_men > 0 else 0

print("% of men who survived:", rate_men)
```

```python
data["Age"] = data["Age"].fillna(data["Age"].median())
data["Adult"] = data["Age"] >= 18
```

```python
total_adults =  data["Adult"].sum()
total_children =   len(data) - total_adults

print("Total number of adults:", total_adults)
print("Total number of children:", total_children)
```

```python
adult_survived = data.loc[data["Adult"] == True, "Survived"]
child_survived = data.loc[data["Adult"] == False, "Survived"]

adult_survival_rate = adult_survived.mean() * 100
child_survival_rate = child_survived.mean() * 100

print(f"Number of adults survived: {adult_survived.sum()}")
print(f"Number of children survived: {child_survived.sum()}")

print(f"Survival rate of adults: {adult_survival_rate:.2f}%")
print(f"Survival rate of children: {child_survival_rate:.2f}%")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
train_data, test_data = train_test_split(data, test_size=0.2, random_state=42)

y_train = train_data["Survived"]
y_test = test_data["Survived"]
features = ["Pclass", "Sex", "SibSp", "Parch"]

X_train = pd.get_dummies(train_data[features])
X_test = pd.get_dummies(test_data[features])

X_test = X_test.reindex(columns=X_train.columns, fill_value=0)


model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy * 100:.2f}%")
```
